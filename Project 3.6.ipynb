```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel("C:/Users/Jeremy/Downloads/Stroke_Prediction_Dataset.xlsx")

df.head()
```

```python

plt.figure(figsize=(8, 5))

plt.hist(df["Age"], bins=20, edgecolor="black")

plt.title("Age Distribution of Patients")
plt.xlabel("Age")
plt.ylabel("Number of Patients")

plt.show()
```

```python
stroke_patients = df[df["Diagnosis"] == "Stroke"]

work_counts = stroke_patients["Work Type"].value_counts()

plt.figure(figsize=(8, 5))
work_counts.plot(kind="bar", edgecolor="black")

plt.title("Work Type Distribution Among Patients Diagnosed with Stroke")
plt.xlabel("Work Type")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    stroke_patients["Age"],
    stroke_patients["Body Mass Index (BMI)"],
    alpha=0.6
)

plt.title("Age vs. BMI Among Patients Diagnosed with Stroke")
plt.xlabel("Age")
plt.ylabel("Body Mass Index (BMI)")
plt.show()
```

```python

numeric_data = df[
    ["Age", "Body Mass Index (BMI)", "Average Glucose Level"]
]

correlation_matrix = numeric_data.corr()

plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```
